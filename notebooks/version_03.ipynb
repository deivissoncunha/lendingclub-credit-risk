# Projeto LendingClub


Para este projeto, estaremos explorando dados disponíveis publicamente de [LendingClub.com](www.lendingclub.com). Lending Club conecta pessoas que precisam de dinheiro (mutuários) com pessoas que têm dinheiro (investidores). Felizmente, como investidor, você gostaria de investir em pessoas que mostraram um perfil de ter uma alta probabilidade de pagá-lo de volta. Vamos tentar criar um modelo que ajude a prever isso.

O clube de empréstimo teve um [ano muito interessante em 2016](https://en.wikipedia.org/wiki/Lending_Club#2016), então vamos verificar alguns de seus dados e ter em mente o contexto. Esses dados são de antes mesmo de serem públicos.

Utilizaremos os dados de empréstimos de 2007-2010 e tentaremos classificar e prever se o mutuário pagou o empréstimo na íntegra. Você pode baixar os dados de [aqui](https://www.lendingclub.com/info/download-data.action) ou apenas usar o csv já fornecido. Recomenda-se que você use o csv fornecido, uma vez que foi limpo dos valores de NA.

Aqui estão o que as colunas representam:
* credit.policy: 1 se o cliente atender aos critérios de subscrição de crédito da LendingClub.com e 0 caso contrário.
* purpose: O objetivo do empréstimo (possíveis valores "credit_card", "debt_consolidation", "educacional", "major_purchase", "small_business" e "all_other").
* int.rate: a taxa de juros do empréstimo (uma taxa de 11% seria armazenada como 0.11). Os mutuários julgados por LendingClub.com para serem mais arriscados recebem taxas de juros mais elevadas.
* installment: as parcelas mensais devidas pelo mutuário se o empréstimo for financiado.
* log.annual.inc: O log natural da renda anual auto-relatada do mutuário.
* dti: Ratio dívida / rendimento do tomador do empréstimo (montante da dívida dividido pela receita anual).
* fico: a pontuação de crédito FICO do mutuário.
* days.with.cr.line: O número de dias em que o mutuário teve uma linha de crédito.
* revol.bal: Saldo rotativo do mutuário (montante não pago no final do ciclo de cobrança do cartão de crédito).
* revol.util: taxa de utilização da linha rotativa do mutuário (o valor da linha de crédito usada em relação ao crédito total disponível).
* inq.last.6mths: número de consultas do mutuário por credores nos últimos 6 meses.
* delinq.2yrs: o número de vezes que o mutuário havia passado mais de 30 dias em um pagamento nos últimos 2 anos.
* pub.rec: O número de registros públicos depreciativos do mutuário (arquivamentos de falências, ônus fiscais ou julgamentos).
* not.fully.paid: empréstimo terminará inadimplente (1) ou será pago integralmente (0).

# Helper Function

```python
def avaliar_modelo(model, X_test, y_test, yhat, nome_modelo='Modelo'):
    
    # ==========================================
    # Probabilidades da classe positiva (1)
    # ==========================================
    
    y_prob = model.predict_proba(X_test)[:, 1]
    
    
    # ==========================================
    # Métricas
    # ==========================================
    
    accuracy = accuracy_score(y_test, yhat)
    
    precision = precision_score(
        y_test,
        yhat,
        zero_division=0
    )
    
    recall = recall_score(
        y_test,
        yhat,
        zero_division=0
    )
    
    f1 = f1_score(
        y_test,
        yhat,
        zero_division=0
    )
    
    roc_auc = roc_auc_score(
        y_test,
        y_prob
    )
    
    pr_auc = average_precision_score(
        y_test,
        y_prob
    )
    
    
    # ==========================================
    # Matriz de confusão
    # ==========================================
    
    cm = confusion_matrix(y_test, yhat)
    
    tn, fp, fn, tp = cm.ravel()
    
    specificity = tn / (tn + fp)
    
    
    # ==========================================
    # Tabela de métricas
    # ==========================================
    
    metricas = pd.DataFrame({
        'Métrica': [
            'Accuracy',
            'Precision',
            'Recall',
            'F1-Score',
            'Specificity',
            'ROC-AUC',
            'PR-AUC'
        ],
        
        'Valor': [
            accuracy,
            precision,
            recall,
            f1,
            specificity,
            roc_auc,
            pr_auc
        ]
    })
    
    metricas['Valor'] = metricas['Valor'].round(4)
    
    
    # ==========================================
    # Exibe tabela
    # ==========================================
    
    print('=' * 50)
    print(f'AVALIAÇÃO: {nome_modelo}')
    print('=' * 50)
    
    display(metricas)
    
    
    # ==========================================
    # Matriz de confusão
    # ==========================================
    
    fig, ax = plt.subplots(figsize=(6, 5))
    
    ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=['Fully Paid (0)', 'Not Fully Paid (1)']
    ).plot(
        ax=ax,
        values_format='d',
        colorbar=False
    )
    
    ax.set_title(f'Matriz de Confusão - {nome_modelo}')
    plt.tight_layout()
    plt.show()
    
    
    # ==========================================
    # Curva ROC
    # ==========================================
    
    fpr, tpr, _ = roc_curve(
        y_test,
        y_prob
    )
    
    fig, ax = plt.subplots(figsize=(7, 5))
    
    ax.plot(
        fpr,
        tpr,
        label=f'{nome_modelo} (AUC = {roc_auc:.3f})'
    )
    
    ax.plot(
        [0, 1],
        [0, 1],
        linestyle='--',
        label='Random'
    )
    
    ax.set_xlabel('False Positive Rate')
    ax.set_ylabel('True Positive Rate')
    ax.set_title(f'Curva ROC - {nome_modelo}')
    ax.legend()
    ax.grid(alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    
    # ==========================================
    # Curva Precision-Recall
    # ==========================================
    
    precision_curve, recall_curve, _ = precision_recall_curve(
        y_test,
        y_prob
    )
    
    baseline = y_test.mean()
    
    fig, ax = plt.subplots(figsize=(7, 5))
    
    ax.plot(
        recall_curve,
        precision_curve,
        label=f'{nome_modelo} (PR-AUC = {pr_auc:.3f})'
    )
    
    ax.axhline(
        baseline,
        linestyle='--',
        label=f'Baseline = {baseline:.3f}'
    )
    
    ax.set_xlabel('Recall')
    ax.set_ylabel('Precision')
    ax.set_title(f'Curva Precision-Recall - {nome_modelo}')
    ax.legend()
    ax.grid(alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    
    # ==========================================
    # Retorno
    # ==========================================
    
    resultado = {
        'Modelo': nome_modelo,
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1,
        'Specificity': specificity,
        'ROC-AUC': roc_auc,
        'PR-AUC': pr_auc,
        'TN': tn,
        'FP': fp,
        'FN': fn,
        'TP': tp
    }
    
    return resultado
```

```python
def calcular_metricas(y_true, y_pred, y_prob):
    
    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred
    ).ravel()

    specificity = tn / (tn + fp)

    return {
        'Accuracy': accuracy_score(y_true, y_pred),
        'Precision': precision_score(
            y_true,
            y_pred,
            zero_division=0
        ),
        'Recall': recall_score(
            y_true,
            y_pred,
            zero_division=0
        ),
        'F1-Score': f1_score(
            y_true,
            y_pred,
            zero_division=0
        ),
        'Specificity': specificity,
        'ROC-AUC': roc_auc_score(
            y_true,
            y_prob
        ),
        'PR-AUC': average_precision_score(
            y_true,
            y_prob
        ),
        'TN': tn,
        'FP': fp,
        'FN': fn,
        'TP': tp
    }
```

```python
def cross_validation(
    X_train,
    y_train,
    model_name,
    model,
    n_splits=5
):
    
    # ==========================================
    # 1. Define o Stratified K-Fold
    # ==========================================
    
    cv = StratifiedKFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=42
    )
    
    
    # ==========================================
    # 2. Lista para armazenar os resultados
    # ==========================================
    
    resultados = []
    
    
    # ==========================================
    # 3. Percorre os folds
    # ==========================================
    
    for fold, (train_idx, val_idx) in enumerate(
        cv.split(X_train, y_train),
        start=1
    ):
        
        # --------------------------------------
        # Separa treino e validação
        # --------------------------------------
        
        X_train_fold = X_train.iloc[train_idx]
        X_val_fold = X_train.iloc[val_idx]
        
        y_train_fold = y_train.iloc[train_idx]
        y_val_fold = y_train.iloc[val_idx]
        
        
        # --------------------------------------
        # Cria uma cópia independente do modelo
        # --------------------------------------
        
        m = clone(model)
        
        
        # --------------------------------------
        # Treinamento
        # --------------------------------------
        
        m.fit(
            X_train_fold,
            y_train_fold
        )
        
        
        # --------------------------------------
        # Previsão da classe
        # --------------------------------------
        
        yhat = m.predict(
            X_val_fold
        )
        
        
        # --------------------------------------
        # Probabilidade da classe 1
        # --------------------------------------
        
        y_prob = m.predict_proba(
            X_val_fold
        )[:, 1]
        
        
        # --------------------------------------
        # Calcula métricas
        # --------------------------------------
        
        metricas = calcular_metricas(
            y_val_fold,
            yhat,
            y_prob
        )
        
        
        # --------------------------------------
        # Adiciona informações do fold
        # --------------------------------------
        
        metricas['Modelo'] = model_name
        metricas['Fold'] = fold
        
        resultados.append(metricas)
    
    
    # ==========================================
    # 4. Converte para DataFrame
    # ==========================================
    
    resultados = pd.DataFrame(resultados)
    
    
    # ==========================================
    # 5. Organiza as colunas
    # ==========================================
    
    colunas = [
        'Modelo',
        'Fold',
        'Accuracy',
        'Precision',
        'Recall',
        'F1-Score',
        'Specificity',
        'ROC-AUC',
        'PR-AUC',
        'TN',
        'FP',
        'FN',
        'TP'
    ]
    
    resultados = resultados[colunas]
    
    
    return resultados
```

```python

def cross_validation_threshold(
    X_train,
    y_train,
    model_name,
    model,
    thresholds=np.arange(0.10, 0.51, 0.05),
    n_splits=5
):

    cv = StratifiedKFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=42
    )

    resultados = []

    for fold, (train_idx, val_idx) in enumerate(
        cv.split(X_train, y_train),
        start=1
    ):

        # -------------------------------
        # Separa treino e validação
        # -------------------------------

        X_train_fold = X_train.iloc[train_idx]
        X_val_fold = X_train.iloc[val_idx]

        y_train_fold = y_train.iloc[train_idx]
        y_val_fold = y_train.iloc[val_idx]

        # -------------------------------
        # Novo modelo
        # -------------------------------

        m = clone(model)

        # -------------------------------
        # Treinamento
        # -------------------------------

        m.fit(
            X_train_fold,
            y_train_fold
        )

        # -------------------------------
        # Probabilidade classe 1
        # -------------------------------

        y_prob = m.predict_proba(
            X_val_fold
        )[:, 1]

        # -------------------------------
        # Testa os thresholds
        # -------------------------------

        for threshold in thresholds:

            yhat = (
                y_prob >= threshold
            ).astype(int)

            tn, fp, fn, tp = confusion_matrix(
                y_val_fold,
                yhat
            ).ravel()

            specificity = tn / (tn + fp)

            resultados.append({
                'Modelo': model_name,
                'Fold': fold,
                'Threshold': threshold,

                'Accuracy': accuracy_score(
                    y_val_fold,
                    yhat
                ),

                'Precision': precision_score(
                    y_val_fold,
                    yhat,
                    zero_division=0
                ),

                'Recall': recall_score(
                    y_val_fold,
                    yhat,
                    zero_division=0
                ),

                'F1-Score': f1_score(
                    y_val_fold,
                    yhat,
                    zero_division=0
                ),

                'Specificity': specificity,

                'TN': tn,
                'FP': fp,
                'FN': fn,
                'TP': tp
            })

    return pd.DataFrame(resultados)
```

```python
def resumo_cv(resultados):

    metricas = [
        'Accuracy',
        'Precision',
        'Recall',
        'F1-Score',
        'Specificity',
        'ROC-AUC',
        'PR-AUC'
    ]

    modelo = resultados['Modelo'].iloc[0]

    resumo = resultados[metricas].agg(['mean', 'std']).T

    resultado = {
        'Modelo': modelo
    }

    for metrica in metricas:
        resultado[f'{metrica}_mean'] = resumo.loc[metrica, 'mean']
        resultado[f'{metrica}_std'] = resumo.loc[metrica, 'std']

    return resultado
```

```python
taxa_renda = (
    df1.groupby('faixa_renda', observed=False)
       .agg(
           clientes=('not.fully.paid', 'count'),
           good=('not.fully.paid', lambda x: (x == 0).sum()),
           bad=('not.fully.paid', lambda x: (x == 1).sum())
       )
       .reset_index()
)

# Distribuição de Good e Bad
taxa_renda['dist_good'] = taxa_renda['good'] / taxa_renda['good'].sum()
taxa_renda['dist_bad'] = taxa_renda['bad'] / taxa_renda['bad'].sum()

# WoE
taxa_renda['WoE'] = np.log(
    taxa_renda['dist_good'] / taxa_renda['dist_bad']
)

# IV por faixa
taxa_renda['IV'] = (
    (taxa_renda['dist_good'] - taxa_renda['dist_bad'])
    * taxa_renda['WoE']
)

taxa_renda.round(4)
```

```python
# crição das faixa (bins)
discrete_bins = {
    'inq.last.6mths': [-1, 0, 1, 2, np.inf],
    'delinq.2yrs': [-1, 0, 1, np.inf],
    'pub.rec': [-1, 0, np.inf]
}
```

```python
def calculate_woe_iv(
    df,
    feature,
    target,
    variable_types,
    discrete_bins,
    n_bins=5
):
    
    data = df[[feature, target]].copy()
    
    variable_type = variable_types[feature]
    
    # -----------------------------
    # Continuous
    # -----------------------------
    
    if variable_type == 'continuous':
        
        data['bin'] = pd.qcut(
            data[feature],
            q=n_bins,
            duplicates='drop'
        )
    
    # -----------------------------
    # Binary / categorical
    # -----------------------------
    
    elif variable_type in ['binary', 'categorical']:
        
        data['bin'] = data[feature].fillna(
            'Missing'
        )
   
    
    # -----------------------------
    # Discrete
    # -----------------------------
    
    elif variable_type == 'discrete':
    
        data['bin'] = pd.cut(
            data[feature],
            bins=discrete_bins[feature]
        )
    
    else:
        raise ValueError(
            f'Tipo desconhecido: {variable_type}'
        )

    # 2. Agrupar Good e Bad
    grouped = (
        data
        .groupby('bin', observed=True)[target]
        .agg(
            total='count',
            bad='sum'
        )
        .reset_index()
    )

    # 3. Calcular Good
    grouped['good'] = grouped['total'] - grouped['bad']

    # 4. Totais
    total_good = grouped['good'].sum()
    total_bad = grouped['bad'].sum()
    grouped['bad_rate'] = (grouped['bad'] / grouped['total']
)

    # 5. Distribuição Good e Bad
    grouped['dist_good'] = grouped['good'] / total_good
    grouped['dist_bad'] = grouped['bad'] / total_bad

    # 6. Evitar divisão por zero
    epsilon = 0.000001

    grouped['dist_good'] = grouped['dist_good'].clip(lower=epsilon)
    grouped['dist_bad'] = grouped['dist_bad'].clip(lower=epsilon)

    # 7. WoE
    grouped['woe'] = np.log(
        grouped['dist_good'] / grouped['dist_bad']
    )

    # 8. IV por bin
    grouped['iv'] = (
        (grouped['dist_good'] - grouped['dist_bad'])
        * grouped['woe']
    )

    # 9. IV total da variável
    iv_total = grouped['iv'].sum()

    # 10. Identificação da variável
    grouped.insert(0, 'feature', feature)

    return grouped, iv_total
```

```python
def calculate_woe_iv_all(
    df,
    target,
    variable_types,
    discrete_bins,
    n_bins=5
):
    
    woe_tables = []
    iv_summary = []
    
    features = [col for col in df.columns if col != target]
    
    for feature in features:
        
        woe_table, iv_total = calculate_woe_iv(
            df=df,
            feature=feature,
            target=target,
            variable_types=variable_types,
            discrete_bins=discrete_bins,
            n_bins=n_bins
        )
        
        woe_tables.append(woe_table)
        
        iv_summary.append({
            'feature': feature,
            'iv': iv_total
        })
    
    
    df_woe_iv = pd.concat(
        woe_tables,
        ignore_index=True
    )
    
    df_iv = (
        pd.DataFrame(iv_summary)
        .sort_values(
            'iv',
            ascending=False
        )
        .reset_index(drop=True)
    )
    
    return df_woe_iv, df_iv
```

## Conclusão

A etapa de otimização de hiperparâmetros foi concentrada na Regressão Logística e no Random Forest. A Regressão Logística foi mantida por representar o modelo tradicional de credit scoring, amplamente utilizado em modelos de Application Score devido à sua interpretabilidade e facilidade de explicação dos coeficientes. O Random Forest foi escolhido como principal benchmark não linear por apresentar bom desempenho inicial, maior capacidade de generalização do que uma árvore única e potencial de melhoria por meio da otimização de hiperparâmetros. Embora o XGBoost também tenha apresentado desempenho competitivo, sua otimização exigiria uma busca em um espaço muito maior de hiperparâmetros, aumentando significativamente o custo computacional. Para o objetivo deste case, optou-se por aprofundar a comparação entre um modelo linear altamente interpretável e um ensemble robusto, mantendo o equilíbrio entre rigor analítico e eficiência experimental.

O XGBoost foi avaliado tanto em sua configuração padrão quanto utilizando scale_pos_weight para tratar o desbalanceamento das classes. Apesar do aumento de Recall em thresholds baixos, o ganho veio acompanhado de redução na Precision, resultando em F1 inferior aos modelos de Regressão Logística e Random Forest. Como não houve ganho incremental relevante e o modelo apresenta maior complexidade de ajuste e manutenção, optou-se por manter apenas os dois modelos com melhor relação entre desempenho e interpretabilidade.

"Eu optei por manter a distribuição original dos dados durante o treinamento porque, em modelos de crédito, gerar exemplos sintéticos pode produzir combinações pouco realistas de variáveis financeiras. Preferi utilizar validação estratificada para garantir estimativas robustas de desempenho e realizar a otimização do threshold para alinhar o modelo ao objetivo de negócio."

### Modelo ajustado 

```python
# --------------------------------------------------
# Pipeline: StandardScaler + Regressão Logística
# --------------------------------------------------

model_lr_tuned = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(
        C=0.01,
        penalty='l2',
        solver='liblinear',
        max_iter=5000,
        random_state=42
    ))
])
        
# training
model_lr_tuned.fit(X_train_enco, y_train_tree)


# -------------------------------
# Probabilidade classe 1
# -------------------------------
y_prob_lr_tuned = model_lr_tuned.predict_proba(
    X_test_enco
)[:, 1]
# -------------------------------
# Testa os thresholds
# -------------------------------

thresholds=np.arange(0.10, 0.51, 0.05)

resultados = []

for threshold in thresholds:
    yhat = (y_prob_lr_tuned >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test_tree,
        yhat
    ).ravel()
    specificity = tn / (tn + fp)

    resultados.append({
        'Threshold': threshold,
        'Accuracy': accuracy_score(y_test_tree, yhat),
        'Precision': precision_score(y_test_tree, yhat, zero_division=0),
        'Recall': recall_score(y_test_tree, yhat, zero_division=0),
        'F1-Score': f1_score(y_test_tree, yhat, zero_division=0),
        'Specificity': specificity,
        'TN': tn,
        'FP': fp,
        'FN': fn,
        'TP': tp
    })
    
df_thresholds = pd.DataFrame(resultados)

df_thresholds
```

```python
def avaliar_thresholds(y_true, y_prob):
    resultados = []

    for threshold in np.arange(0.10, 0.51, 0.05):

        y_pred = (y_prob >= threshold).astype(int)

        tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()

        resultados.append({
            "Threshold": threshold,
            "Accuracy": accuracy_score(y_true, y_pred),
            "Precision": precision_score(y_true, y_pred, zero_division=0),
            "Recall": recall_score(y_true, y_pred, zero_division=0),
            "F1-Score": f1_score(y_true, y_pred, zero_division=0),
            "Specificity": tn/(tn+fp),
            "TN": tn,
            "FP": fp,
            "FN": fn,
            "TP": tp
        })

    return pd.DataFrame(resultados)
```

## Comparação dos modelos usando a Estatística KS

```python
def calcular_ks(y_true, probas):

    fpr, tpr, thresholds = roc_curve(y_true, probas)

    ks = np.max(tpr - fpr)

    idx = np.argmax(tpr - fpr)

    return ks, thresholds[idx], tpr[idx], fpr[idx]
```

```python
ks_lr = calcular_ks(y_test_tree, y_prob_lr_tuned)
ks_rf = calcular_ks(y_test_tree, y_prob_rf_tuned)

print(ks_lr)
print(ks_rf)
```

Eu manteria a Regressão Logística como baseline interpretável, mas passaria a considerar o Random Forest como o modelo final do Application Score, justamente porque ele combina:

desempenho competitivo em F1;

possibilidade de explicar as variáveis mais importantes com Permutation Importance.

Em um projeto de Credit Scoring, eu daria um peso maior ao KS do que ao F1, porque o objetivo principal não é apenas classificar corretamente, mas ordenar os clientes do menor para o maior risco.

O Random Forest apresentou um KS cerca de 2,8 pontos percentuais maior que a Regressão Logística (27,4% vs. 24,5%), indicando melhor capacidade de separar bons e maus pagadores ao longo da distribuição de risco. Embora a Regressão Logística tenha obtido um F1 ligeiramente superior, a diferença foi pequena, enquanto o ganho em poder de discriminação favorece o Random Forest.

## Credit Score

```python
def credit_score(prob_default,
                 base_score=600,
                 base_odds=20,
                 pdo=50):

    factor = pdo / np.log(2)
    offset = base_score - factor*np.log(base_odds)

    odds = (1-prob_default)/prob_default

    score = offset + factor*np.log(odds)

    return np.round(score).astype(int)
```

A ideia é transformar a matriz de confusão em impacto financeiro.

Por exemplo:

FP (rejeitar um bom cliente): custo = R$500.

FN (aprovar um inadimplente): custo = R$5.000.

Então calcularíamos:

Custo=500×FP+5000×FN

e plotaríamos esse custo para cada threshold da Regressão Logística e da Random Forest na mesma figura.

Esse tipo de gráfico é extremamente próximo do que acontece em bancos e fintechs: a decisão final deixa de ser "qual tem maior F1?" e passa a ser "qual threshold minimiza o prejuízo esperado?". Essa costuma ser uma conclusão muito forte para encerrar um case de classificação em Ciência de Dados.
